## Problem Three: Missing Number in Range
**Description**  
You are given a list containing `n` non-negative integers in the range `[1, n]`. Return a list of numbers that are missing from the range `[1, n]` in the given list. If there is no missing number, return -1. 

**Note**: integers in the list may not be unique.

**Examples**

* Example 1:

    Input: `lst = [1, 1, 3]`

    Output: [2]

* Example 2:

    Input: `lst = [5,3,2,4,1]`

    Output: -1

* Example 3:

    Input: `lst = [6, 8, 2, 3, 5, 7, 1, 1, 2, 10]`

    Output: [4, 9]

**Problem 3 Starter Code**

```python
from typing import List
def missing_num(nums: List) -> int:
  # Code your solution here #
  
    n = len(nums)
    num_set = set(nums)
    
    # Check range 
    missing = [i for i in range(1, n + 1) if i not in num_set]
    
    return missing if missing else -1

    
# testing example 1: 
num_set = [1, 1, 3]
output = missing_num(num_set)
print(f"Missing number in the example 1 set: {output}")

# testing example 2:
num_set = [5, 3, 2, 4, 1]
output = missing_num(num_set)
print(f"Missing number in the example 2 set: {output}")

# testing example 3:
num_set = [6, 8, 2, 3, 5, 7, 1, 1, 2, 10]
output = missing_num(num_set)
print(f"Missing number in the example 3 set: {output}")
```

```python
# 
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1

print(first_duplicate([1, 2, 3, 2, 4, 4, 5])) 

# My answer:
# With regards to the time complexity, the solution has a time complexity of O(n). This is because he is iterating through the list once and checking if each number is in the seen set, which takes O(1) time on average. 
# The space complexity is also O(n) because in the worst case, all numbers are unique and we need to store them in the seen set.
```
